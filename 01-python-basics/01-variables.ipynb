Ejercicio 3 — zip para calcular

unidades = [3, 5, 2, 8]
precio_unitario = [10.0, 5.5, 20.0, 3.25]

Calcula el total de cada línea multiplicando unidades por precio e imprime el total acumulado al final.

```python
total = 0
unidades = [3, 5, 2, 8]
precio_unitario = [10.0, 5.5, 20.0, 3.25]

for unidad, precio in zip(unidades, precio_unitario):
    total += unidad * precio
    print(f"{total} = {unidad} * {precio} ")
print(f"Total acumulado: {total}")
```
