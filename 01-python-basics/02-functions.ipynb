Crea una función llamada calcular_iva que reciba un precio y devuelva el precio con el 21% de IVA aplicado. Imprímelo fuera de la función.

```python
def calcular_iva(precio):
    iva = precio * 0.21
    precio_con_iva = precio + iva
    return precio_con_iva
calcular_iva(120)
```

```python
def cuadrado (numero):
    """Realiza el cuadrado del numero seleccionado 

    Args:
        numero (int, float): Los números para calcular el cuadrado

    Returns:
        int, float: El cuadrado del número ingresado
    """
    return numero ** 2
print(cuadrado (5))
print(cuadrado(3.7))
print(cuadrado(200))
```

```python
def calcular_descuento (precio, porcentaje):
    """Calcula el descuento de un precio

    Args:
        precio (int,float): Es el precio del producto
        porcentaje (int): Es el porcentaje de descuento que se le aplicará al precio

    Returns:
        int,float: El precio con el descuento aplicado
    """
    return precio - (precio * porcentaje / 100)
print(f"El precio final es --> {calcular_descuento(100, 20)}")
print (f"El precio final es --> {calcular_descuento(50, 50)}")
print (f"El precio final es --> {calcular_descuento(200, 10)}")
```

```python
def clasificar_temperatura (temperatura):
    if temperatura < 10:
        return "Frío"
    elif temperatura <= 25:
        return "Templado"
    else:
        return "Calor"

print(clasificar_temperatura(5))
print(clasificar_temperatura(15))
print(clasificar_temperatura(30))
```

```python
def calcular_precio(precio, descuento = 10):
    precio_final = precio - (precio * descuento/100)
    return precio_final
print(calcular_precio(100))
print(calcular_precio(120, 0))
```

```python
def crear_usuario(nombre, rol="usuario", activo=True):
    usuario_creado = nombre + " - " + rol + " - " + str(activo)
    return usuario_creado
print(crear_usuario("Juan"))
print(crear_usuario("Ana", "usuario"," "))
print(crear_usuario("Alba", " "," "))
```

```python
def sumar_todo (*args):
    total = 0
    for numero in args:
        total += numero
    return total
print(sumar_todo(3, 5, 1))
print(sumar_todo(10, 20, 30, 40, 50))
print(sumar_todo(7))          
```

```python
def mayor_numero (*numero):
    return max(numero)
print(mayor_numero(1,8,34,2))
```

```python
def mostrar_info(**Informacion):
    for clave, valor in Informacion.items():
        print(f"{clave} ----> {valor}")
mostrar_info(nombre="Juan", edad=30, ciudad="Madrid", CP=28001 )
```

```python
def suma_hasta (numero):
    total = 0
    for i in range(1, numero + 1):
        total += i
    return total
print(suma_hasta(5))
```
