```python
class Producto:
    def __init__(self, Nombre, Precio, Stock):
        self.nombre = Nombre
        self.precio = Precio
        self.stock = Stock
    def vender(self, Cantidad):
        self.cantidad = Cantidad
        self.stock -= self.cantidad
        if self.stock <= 0:
            print(f"El producto {self.nombre} se ha agotado.")
    def valor_total(self):
        return self.precio * self.stock
    def mostrar_info(self):
        print(f"Producto: {self.nombre}, Precio: {self.precio}, Stock: {self.stock}")

Producto1 = Producto("PlayStation 5", 499.99, 10)
Producto1.mostrar_info()
Producto1.vender(8)
Producto1.valor_total()
print(Producto1.valor_total())
Producto1.vender(3)
```
