Ejercicio: Pide (simula) que un precio llega como texto "49.99" desde una API. Conviértelo a float y súmale un 21% de IVA.

```python
precio_API = "49.99"
print(type(precio_API))

precio_numero = float(precio_API)
print(type(precio_numero))

IVA = precio_numero * 0.21

precio_final = precio_numero + IVA
print(f"El prrecio final del producto es: {precio_final:.2f}")
```

Ejercicio: Dado mensaje = "  Hola Cliente!  ", elimina los espacios sobrantes, ponlo en mayúsculas, y muestra solo los primeros 4 caracteres.

Ejercicio: Tienes productos = ["Teclado", "Ratón"] y precios = [45, 20]. Usa enumerate() para imprimir "1. Teclado - 45€", "2. Ratón - 20€".

```python
productos = ["Teclado", "Ratón"]
precios = [45, 20]

for producto, precio in zip(productos, precios):
    print (producto, precio)
```

```python
for indice, (producto, precio) in enumerate(zip(productos, precios), start=1):
    print(indice, producto, precio)
```
